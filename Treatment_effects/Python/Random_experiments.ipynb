```python
#3.14.2
import pandas as pd
import numpy as np
import pyreadstat
from pathlib import Path
import scipy.stats as stats
import statsmodels.api as sms

#Three treatment groups, assigned randomly.
df = pd.read_stata("C:/Causal_Inference/Treatment_effects/data/chp2.dta")
```

```python
#Grouped summary stats
means = df.groupby('Txgroup')['outcome'].describe()
print(means)

#Confidence Intervals by Tx group
def mean_ci(series, confidence=0.95):
    mean = series.mean()
    sem = stats.sem(series)
    MoE = sem * stats.t.ppf((1 + confidence) / 2, len(series) - 1)
    return pd.Series({'mean': mean, 
                      'ci_lower': mean - MoE, 
                      'ci_upper': mean + MoE})

ci_means = df.groupby('Txgroup')['outcome'].apply(mean_ci)
print(ci_means)
```

```python
#T-test between Txgroup 1 and 2
online = df[df['Txgroup'] == "online"]['outcome']
F2F = df[df['Txgroup'] == "face-to-face"]['outcome']

ttest_result = stats.ttest_ind(online, F2F, equal_var=False)  # Welch's t-test
print(f"T-statistic: {ttest_result.statistic}, p-value: {ttest_result.pvalue}")
```
