```python
import statsmodels.api as sm

# Load the data
df = pd.read_stata("/Causal_Inference/Treatment_effects/data/cattaneo2.dta",
                      convert_categoricals=False)
#print(df.head(10))

# Regression for control group (mbsmoke == 0)
control_data = df[df['mbsmoke'] == 0]
X_control = sm.add_constant(control_data['mage'])  # Add intercept
y_control = control_data['bweight']
model_control = sm.OLS(y_control, X_control).fit()
print(model_control.summary())

# Predicted values for control group (yhat0)
df['yhat0'] = 3108.28 + (11.36 * df['mage'])
print(df['yhat0'].mean())

# Regression for treatment group (mbsmoke == 1)
treatment_data = df[df['mbsmoke'] == 1]
X_treatment = sm.add_constant(treatment_data['mage'])  # Add intercept
y_treatment = treatment_data['bweight']
model_treatment = sm.OLS(y_treatment, X_treatment).fit()
print(model_treatment.summary())

# Predicted values for treatment group (yhat1)
df['yhat1'] = 3237.091 + (-3.951 * df['mage'])
print(df['yhat1'].mean())
# Calculate the difference
df['diff'] = df['yhat1'] - df['yhat0']

#Average treatment effect (ATE)
print(df['diff'].mean())

#New tibble equivalent: summary with mean diff (ATE)
ate_summary = df[['diff']].mean()
print('ATE:', ate_summary)

#Average treatment effect for the treated (ATET)
atet = df.groupby('mbsmoke')['diff'].mean()
print('ATET:', atet)


```

```python
import statsmodels.api as sm

# Load the data
df = pd.read_stata("/Causal_Inference/Treatment_effects/data/cattaneo2.dta",
                      convert_categoricals=False)

# Regression for control group (mbsmoke == 0)
control_data = df[df['mbsmoke'] == 0]
X_control = sm.add_constant(control_data[['fage', 'mage', 'mmarried', 'prenatal1']])  # Add intercept
y_control = control_data['bweight']
model_control = sm.OLS(y_control, X_control).fit()
print(model_control.summary())

# Predicted values for control group (yhat0)
df['yhat0'] = (3120.4836) + (-0.3222 * df['fage']) + \
                 (4.9787 * df['mage']) + (162.2292 * df['mmarried']) + (55.6804 * df['prenatal1'])
print(df['yhat0'].mean())

# Regression for treatment group (mbsmoke == 1)
treatment_data = df[df['mbsmoke'] == 1]
X_treatment = sm.add_constant(treatment_data[['fage', 'mage', 'mmarried', 'prenatal1']])  # Add intercept
y_treatment = treatment_data['bweight']
model_treatment = sm.OLS(y_treatment, X_treatment).fit()
print(model_treatment.summary())

# Predicted values for treatment group (yhat1)
df['yhat1'] = (3267.10555) + (-0.03353 * df['fage']) + \
                 (-8.51209 * df['mage']) + (131.72129 * df['mmarried']) + (33.70559 * df['prenatal1'])
print(df['yhat1'].mean())

# Calculate the difference between predicted values (yhat1 - yhat0)
df['diff'] = df['yhat1'] - df['yhat0']

# Average treatment effect (ATE)
ate = df['diff'].mean()
print(f"ATE: {ate}")

# New tibble (summary with mean diff - ATE)
ate_summary = df[['diff']].mean()
print(ate_summary)

# Average treatment effect for the treated (ATET) - Group by 'mbsmoke'
atet = df.groupby('mbsmoke')['diff'].mean()
print(atet)
```
